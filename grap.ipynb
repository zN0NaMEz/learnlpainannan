```python
import matplotlib.pyplot as plt

users = [1000, 2000, 6000]
revenue = [6564000, 11712000, 36552000]
years = ['ปีที่ 1', 'ปีที่ 2', 'ปีที่ 3']

plt.figure(figsize=(9, 5))
plt.plot(users, revenue, marker='o', linewidth=2.5, color='#2563EB')

for x, y, label in zip(users, revenue, years):
    plt.annotate(
        f'{label}\n{x:,} users\n฿{y:,.0f}',
        (x, y),
        textcoords='offset points',
        xytext=(0, 12),
        ha='center'
    )

plt.title('การเติบโตของจำนวนผู้ใช้และรายได้รวมใน 3 ปี')
plt.xlabel('จำนวนผู้ใช้สะสมสิ้นปี (คน)')
plt.ylabel('รายได้รวม (บาท)')
plt.grid(True, linestyle='--', alpha=0.4)
plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```
